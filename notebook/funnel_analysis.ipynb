```python
poutcome_conversion = pd.crosstab(df['poutcome'], df['y'], normalize='index') * 100

poutcome_conversion['yes'].plot(kind='bar', figsize=(8,5))

plt.title("Conversion Rate by Previous Campaign Outcome")
plt.ylabel("Conversion %")
plt.xlabel("Previous Outcome")

plt.show()
```

```python
job_conversion = pd.crosstab(df['job'], df['y'], normalize='index') * 100

job_conversion['yes'].sort_values().plot(
    kind='bar',
    figsize=(12,6)
)

plt.title("Conversion Rate by Job Type")
plt.xlabel("Job")
plt.ylabel("Conversion Rate (%)")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x='y', data=df)

plt.title("Customer Conversion Distribution")
plt.xlabel("Subscription")
plt.ylabel("Number of Customers")

plt.savefig("../dashboard/conversion_distribution.png", dpi=300, bbox_inches='tight')

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x='contact', hue='y', data=df)

plt.title("Conversion by Contact Method")
plt.xlabel("Contact Type")
plt.ylabel("Customers")

plt.savefig("../dashboard/contact_conversion.png", dpi=300, bbox_inches='tight')

plt.show()
```

```python
plt.figure(figsize=(8,4))

sns.countplot(x='month', hue='y', data=df)

plt.title("Conversion by Month")
plt.xlabel("Month")
plt.ylabel("Customers")

plt.savefig("../dashboard/month_conversion.png", dpi=300, bbox_inches='tight')

plt.show()
```

```python
job_conversion = pd.crosstab(df['job'], df['y'], normalize='index') * 100

job_conversion['yes'].sort_values().plot(
    kind='bar',
    figsize=(10,5)
)

plt.title("Conversion Rate by Job Type")
plt.xlabel("Job")
plt.ylabel("Conversion Rate (%)")

plt.savefig("../dashboard/job_conversion_rate.png", dpi=300, bbox_inches='tight')

plt.show()
```
